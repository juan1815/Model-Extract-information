```python
import pandas as pd
import json

# 1. Leer el JSON con los proyectos extraídos
with open('novedades_2019_2020.json', 'r', encoding='utf-8') as f:
    novedades = json.load(f)

# 2. Convertirlo en un DataFrame plano con 'numeroProyecto' y 'Estado'
rows = []
# Para cada proyecto archivado
for item in novedades.get('Archivados', []):
    code = item['Código']
    rows.append({'NumeroProyecto': code, 'Estado': 'Archivado'})
# Para cada proyecto retirado
for item in novedades.get('Retirados', []):
    code = item['Código']
    rows.append({'NumeroProyecto': code, 'Estado': 'Retirado'})

estado_df = pd.DataFrame(rows)

# 3. Cargar el Excel original
df = pd.read_excel('2019_2020.xlsx')

# 4. Asegurarnos de que los números de proyecto coincidan en formato
#    (por ejemplo, relleno de ceros a la izquierda si es necesario)
df['NumeroProyecto'] = df['NumeroProyecto'].astype(str).str.zfill(3)
estado_df['NumeroProyecto'] = estado_df['NumeroProyecto'].astype(str).str.zfill(3)

# 5. Hacer el merge para añadir la columna 'Estado'
df_con_estado = df.merge(
    estado_df[['NumeroProyecto', 'Estado']],
    on='NumeroProyecto',
    how='left'    # los proyectos que no estén en el JSON quedarán con NaN
)

# 6. Opcional: si quieres reemplazar los NaN por un estado genérico:
df_con_estado['Estado'] = df_con_estado['Estado'].fillna('Sin novedad')

# 7. Guardar el resultado en un nuevo Excel
output_path = '2019_2020_con_estado.xlsx'
df_con_estado.to_excel(output_path, index=False)

print(f'✅ Se ha creado el archivo con la columna Estado en: {output_path}')
```
